# Phase 2 — Data Validation

### Checking if there are any impossible values such as negative prices or negative quantity

```python
print(f"negative inventory levels: {int((df['Inventory Level'] < 0).sum())}")

print(f"negative units sold: {int((df['Units Sold'] < 0).sum())}")

print(f"negative units ordered: {int((df['Units Ordered'] < 0).sum())}")

print(f"negative price: {int((df['Price'] < 0).sum())}")
```

### Checking if units sold is mentioned more than the inventory levels

```python
print(f"units sold greater than inventory levels: {int((df['Units Sold'] > df['Inventory Level']).sum())}")
```
